```python
import functions_framework
import pandas as pd
import requests
from keys import RapidAPI_key, SQL_PW
from datetime import date, timedelta

@functions_framework.http
def get_arrivals(request):

    from sqlalchemy import create_engine, text

    schema = "cities_info"
    host = "34.14.93.18"
    user = "root"
    password = SQL_PW
    port = 3306

    engine = create_engine(
        f"mysql+pymysql://{user}:{password}@{host}:{port}/{schema}",
        pool_pre_ping=True,
        pool_recycle=280
    )


    tomorrow_str = (date.today() + timedelta(days=1)).strftime("%Y-%m-%d")

    headers = {
        "x-rapidapi-key": RapidAPI_key,
        "x-rapidapi-host": "aerodatabox.p.rapidapi.com"
    }

    all_arrivals = []
    city = ["BER", "HAM", "MUC"]
    timeframes = [
        f"{tomorrow_str}T00:00/{tomorrow_str}T11:59",
        f"{tomorrow_str}T12:00/{tomorrow_str}T23:59",
    ]

    for c in city:
        for t in timeframes:
            url = f"https://aerodatabox.p.rapidapi.com/flights/airports/iata/{c}/{t}"
            response = requests.get(url, headers=headers)

            if response.status_code != 200:
                continue

            data = response.json()
            arrivals_raw = data.get("arrivals", [])

            if not arrivals_raw:
                continue

            df = pd.json_normalize(arrivals_raw)

            df["movement.scheduledTime.local"] = pd.to_datetime(
                df["movement.scheduledTime.local"]
            ).dt.strftime("%Y-%m-%d %H:%M")

            df = pd.DataFrame({
                "arrival_iata": c,
                "flight_num": df.get("number"),
                "arrival_time": df.get("movement.scheduledTime.local"),
                "arrival_terminal": df.get("movement.terminal"),
                "departure_iata": df.get("movement.airport.iata"),
            })

            all_arrivals.append(df)

    if not all_arrivals:
        return ("No data", 204)

    result = pd.concat(all_arrivals, ignore_index=True)

    records = result.to_dict(orient="records")

    insert_sql = text("""
        INSERT INTO flights (
            arrival_iata,
            flight_num,
            arrival_time,
            arrival_terminal,
            departure_iata
        )
        VALUES (
            :arrival_iata,
            :flight_num,
            :arrival_time,
            :arrival_terminal,
            :departure_iata
        )
    """)

    with engine.begin() as conn:
        conn.execute(insert_sql, records)

    return {
        "status": "success",
        "inserted_rows": len(records)
    }
```
