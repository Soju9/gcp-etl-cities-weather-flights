```python
population_df = pd.DataFrame({"population_size": population,
                              "timestamp_population": timestamp})

population_df["city_id"] = pd.array(range(1, len(population_df) + 1), dtype="int32")

population_df
```

```python
population_df.to_sql('population',
                  if_exists='append',
                  con=connection_string,
                  index=False)
```

```python
from sqlalchemy import create_engine, text

# Add foreign key constraints to establish table relationships
engine = create_engine(connection_string)

sql_commands = [
    "ALTER TABLE population ADD CONSTRAINT fk_population_city FOREIGN KEY (city_id) REFERENCES city(city_id)",
    "ALTER TABLE coordinates ADD CONSTRAINT fk_coordinates_city FOREIGN KEY (city_id) REFERENCES city(city_id)"
]

with engine.connect() as conn:
    for command in sql_commands:
        try:
            conn.execute(text(command))
            print(f"âœ“ Success: {command}")
        except Exception as e:
            error_msg = str(e)
            # Ignore errors about duplicate keys/constraints
            if "Duplicate" in error_msg or "already exists" in error_msg:
                print(f"âš  Already exists: {command}")
            else:
                print(f"âœ— Command failed: {command}")
                print(f"  Error: {e}")
    conn.commit()
```
