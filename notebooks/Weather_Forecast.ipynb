```python

def get_weather_forecast(cities, api_key=APIkey):
    all_forecasts = []
    berlin_timezone = timezone('Europe/Berlin')
    
    for city_name, lat, lon in cities:
        # Get weather forecast
        url = f"https://api.openweathermap.org/data/2.5/forecast?lat={lat}&lon={lon}&appid={api_key}&units=metric"
        response = requests.get(url)
        weather_json = response.json()
        
        # Convert to DataFrame
        df = pd.DataFrame(weather_json['list'])
        df = df.drop(columns=['sys', 'clouds', 'visibility'])
        
        # Parse datetime
        df['datetime'] = pd.to_datetime(df['dt'], unit='s')
        daily_forecast = df.copy()
        daily_forecast = daily_forecast.drop(columns=['dt', 'dt_txt'])
        daily_forecast = daily_forecast.reset_index(drop=True)
        
        # Extract nested columns
        main_data = pd.DataFrame(daily_forecast['main'].tolist()).reset_index(drop=True)
        weather_data = daily_forecast['weather'].apply(lambda x: x[0] if isinstance(x, list) and len(x) > 0 else {})
        weather_df = pd.DataFrame(weather_data.tolist()).reset_index(drop=True)
        wind_data = pd.DataFrame(daily_forecast['wind'].tolist()).reset_index(drop=True)

        retrieval_time = datetime.now(berlin_timezone).strftime("%Y-%m-%d")
        
        # Create formatted forecast for this city
        city_forecast = pd.DataFrame({
            'city_id': city_name,
            'Date': daily_forecast['datetime'].dt.strftime('%Y-%m-%d'),
            'Time': daily_forecast['datetime'].dt.strftime('%H:%M'),
            'Temperature_Â°C': main_data['temp'].round(1),
            'Feels_Like_Â°C': main_data['feels_like'].round(1),
            'Min_Temp_Â°C': main_data['temp_min'].round(1),
            'Max_Temp_Â°C': main_data['temp_max'].round(1),
            'Rain_Probability_perc': (daily_forecast['pop'] * 100).astype(int),
            'Humidity_perc': main_data['humidity'],
            'Weather': weather_df['main'],
            'Wind_Speed': wind_data['speed'].round(1),
            'data_retrieved_at': retrieval_time
        })
        
        all_forecasts.append(city_forecast)
    
    # Combine all city forecasts
    result = pd.concat(all_forecasts, ignore_index=True)
    result.index = range(1, len(result) + 1)
    
    return result
```

```python
cities_list = [
    (1, 52.52, 13.405),
    (2, 53.55, 10),
    (3, 48.1375, 11.575)
]

weather_df = get_weather_forecast(cities_list)
weather_df
```
