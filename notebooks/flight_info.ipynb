```python
import pandas as pd
import requests
from sqlalchemy import create_engine, text
```

```python
# Create a function to pull the data for airports from Aerodatabox API for different latitudes and longitudes

from wsgiref import headers


def get_airports(latitudes, longitudes):
    headers = {
        "x-rapidapi-key": "YOUR_RAPIDAPI_KEY",
	    "x-rapidapi-host": "aerodatabox.p.rapidapi.com",
	    "Content-Type": "application/json"
    }
    
    all_airports = []
    
    for lat, lon in zip(latitudes, longitudes):
        url = "https://aerodatabox.p.rapidapi.com/airports/search/location"
        querystring = {"lat":lat,"lon":lon,"radiusKm":"50","limit":"10","withFlightInfoOnly":"true"}
        
        response = requests.get(url, headers=headers, params=querystring)

        if response.status_code == 200:
            data = response.json()
            airports = pd.json_normalize(data.get('items', []))
            all_airports.append(airports)

    return pd.concat(all_airports, ignore_index=True)
```

```python
latitudes = [52.52, 53.55, 48.13]  # Berlin, Hamburg, Munich
longitudes = [13.40, 10.0, 11.58]  # Berlin, Hamburg, Munich

airports = get_airports(latitudes, longitudes)
airports
```

```python
city_ids = {
    "Berlin": 1,
    "Hamburg": 2,
    "Munich": 3,}

airports_df = pd.DataFrame({
                "iata": airports.get("iata"),
                "icao": airports.get("icao"),
                "airport_name": airports.get("name"),
                "city": airports.get("municipalityName"),
                "country": airports.get("countryCode"),
                "latitude": airports.get("location.lat"),
                "longitude": airports.get("location.lon"),
                "city_id": airports.get("municipalityName").map(city_ids)
            })
airports_df
```

```python
from datetime import date, timedelta
tomorrow_str = (date.today() + timedelta(days=1)).strftime("%Y-%m-%d")

def get_arrivals(city):
    headers = {
        "x-rapidapi-key": "YOUR_RAPIDAPI_KEY",
        "x-rapidapi-host": "aerodatabox.p.rapidapi.com",
        "Content-Type": "application/json"
    }

    all_arrivals = []

    if isinstance(city, str):
        city = [city]

    timeframes = [
        f"{tomorrow_str}T00:00/{tomorrow_str}T11:59",
        f"{tomorrow_str}T12:00/{tomorrow_str}T23:59",
    ]

    querystring = {
        "withLeg": "false",
        "direction": "Arrival",
        "withCancelled": "false",
        "withCodeshared": "false",
        "withCargo": "false",
        "withPrivate": "false",
        "withLocation": "false"
    }

    for c in city:
        for t in timeframes:
            url = f"https://aerodatabox.p.rapidapi.com/flights/airports/iata/{c}/{t}"

            response = requests.get(url, headers=headers, params=querystring)

            if response.status_code != 200:
                print(f"Failed: {c} {t} ({response.status_code})")
                continue

            data = response.json()
            arrivals_raw = data.get("arrivals", [])

            if not arrivals_raw:
                print(f"No data for {c} {t}")
                continue

            arrivals = pd.json_normalize(arrivals_raw)
            arrivals["movement.scheduledTime.local"] = pd.to_datetime(
                arrivals["movement.scheduledTime.local"]
                ).dt.strftime("%Y-%m-%d %H:%M")


            arrivals = pd.DataFrame({
                "arrival_iata": c,
                "flight_num": arrivals.get("number"),
                "arrival_time": arrivals.get("movement.scheduledTime.local"),
                "arrival_terminal": arrivals.get("movement.terminal"),
                "departure_iata": arrivals.get("movement.airport.iata"),
            })

            all_arrivals.append(arrivals)

    if not all_arrivals:
        return pd.DataFrame()

    return pd.concat(all_arrivals, ignore_index=True)
```

```python
city = ["BER", "HAM", "MUC"]

arrivals_df = get_arrivals(city)
arrivals_df
```

```python
city = ["CGN", "FMO", "TXL"]

get_arrivals(city)
```
